# Criminaliteit in de VS: oost vs west
Criminaliteit is een van de meest besproken maatschappelijke vraagstukken in grote steden. Factoren zoals bevolkingsgroei, economische ongelijkheid, toerisme en stedelijke ontwikkeling kunnen invloed hebben op de aard en omvang van criminaliteit. Hierdoor kunnen criminaliteitspatronen sterk verschillen tussen steden, maar ook tussen wijken binnen dezelfde stad.

In deze data story worden criminaliteitspatronen in twee van de grootste steden van de Verenigde Staten, New York en Los Angeles, met elkaar vergeleken. De analyse is gebaseerd op uitgebreide politiedatasets met informatie over misdrijven, arrestaties, locaties en betrokken personen. Deze gegevens bieden de mogelijkheid om inzicht te krijgen in de verspreiding van criminaliteit en de factoren die hieraan ten grondslag liggen.

De analyse richt zich op twee centrale vragen. Ten eerste wordt onderzocht of New York meer criminaliteit kent dan Los Angeles en welke stedelijke factoren, zoals bevolkingsdichtheid, toerisme en sociaaleconomische omstandigheden, mogelijke verschillen kunnen verklaren. Ten tweede wordt gekeken naar de mate waarin criminaliteit binnen de steden zelf varieert, met bijzondere aandacht voor verschillen tussen wijken en stadsdelen en de invloed van lokale sociaaleconomische omstandigheden.

Door gebruik te maken van gedetailleerde gegevens over incidenten, arrestaties, locaties en betrokken personen wordt niet alleen gekeken naar gemiddelde criminaliteitscijfers, maar ook naar onderliggende patronen. Op deze manier ontstaat een genuanceerder beeld van criminaliteit binnen en tussen beide steden.
