# Inleiding
Criminaliteit is een van de meest besproken maatschappelijke vraagstukken. Factoren zoals bevolkingsgroei, economische ongelijkheid, toerisme en stedelijke ontwikkeling kunnen invloed hebben op de aard en omvang van criminaliteit. Hierdoor kunnen criminaliteitspatronen sterk verschillen tussen steden, maar ook tussen wijken binnen dezelfde stad.

Het vergelijken van wereldsteden zoals New York City en Los Angeles biedt een interessant perspectief op deze verschillen. Beide behoren tot de grootste en meest invloedrijke steden van de Verenigde Staten, met miljoenen inwoners, een sterke economische positie en een grote culturele diversiteit. Tegelijkertijd verschillen ze sterk in stedelijke structuur, bevolkingsdichtheid en sociale dynamiek. Deze contrasten maken het mogelijk om te onderzoeken hoe stedelijke omstandigheden samenhangen met criminaliteitspatronen.

In deze data story worden criminaliteitspatronen in New York City en Los Angeles met elkaar vergeleken. De analyse is gebaseerd op uitgebreide politiedatasets met informatie over misdrijven, arrestaties en locaties. Deze gegevens bieden de mogelijkheid om inzicht te krijgen in de verspreiding van criminaliteit en de factoren die hieraan ten grondslag liggen.

De analyse richt zich op twee centrale perspectieven. 
1. Hoeveelheden criminaliteit zijn niet hoger in New York City dan in Los Angeles.
    - Totale hoeveelheid criminaliteit tussen NYC en LA is per jaar niet hoger dan bij NYC vergeleken met LA.
    - Totale hoeveelheid criminaliteit tussen NYC en LA is per maand niet hoger dan bij NYC vergeleken met LA.

2. Er zijn verschillen in de soorten criminaliteit tussen NYC en LA
    - Er is een verschil tussen de frequentie van soorten criminaliteit tussen NYC en LA.
    - Er is een verschil tussen de frequentie van soorten criminaliteit binnen NYC.
    - Er is een verschil tussen de frequentie van soorten criminaliteit binnen LA.

Door gebruik te maken van gedetailleerde gegevens over incidenten, arrestaties en locaties wordt niet alleen gekeken naar gemiddelde criminaliteitscijfers, maar ook naar onderliggende patronen. Op deze manier ontstaat een genuanceerder beeld van criminaliteit binnen en tussen beide steden.
