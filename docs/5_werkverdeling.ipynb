# Werkverdeling
De werkzaamheden binnen het project zijn verdeeld op basis van individuele sterke punten. Nina heeft zich voornamelijk gericht op het schrijven van de data story, het formuleren van de argumentatie, het verwerken van feedback en het structureren van de verschillende onderdelen van het verslag. Lisa en Yifan hebben zich beziggehouden met het programmeren, het verwerken van de datasets en het ontwikkelen van de visualisaties.
Hoewel ieder teamlid een eigen focus had, werden belangrijke keuzes gezamenlijk besproken. Ook de voorbereiding van de presentatie vond gezamenlijk plaats en alle teamleden waren aanwezig tijdens het feedback- en presentatiemoment. Hierdoor konden zowel de technische als de inhoudelijke aspecten van het project gezamenlijk worden toegelicht en verbeterd.
