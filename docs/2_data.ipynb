# Datasetbeschrijving 

Voor dit onderzoek is gebruikgemaakt van twee datasets.

## Los Angeles

Voor dit onderzoek is gebruikgemaakt van de dataset **"Crime Data from 2020 to 2024"**, beschikbaar gesteld door de Los Angeles Police Department (LAPD). De dataset bevat informatie over geregistreerde misdrijven in de stad Los Angeles tussen 2020 en 2024. Elke rij vertegenwoordigt één geregistreerd misdrijf. De dataset bevat ongeveer **1 miljoen observaties (rijen)** en **28 variabelen (kolommen)**.

### Variabelen in de dataset

#### 1. Identificatie van het misdrijf

Deze variabelen identificeren een specifiek politieverslag:

- **dr_no**: uniek dossiernummer van het misdrijf.
- **part_1_2**: classificatie van het misdrijf volgens de FBI-indeling (*Part I* of *Part II*).

#### 2. Tijdsgegevens

Deze variabelen beschrijven wanneer het misdrijf plaatsvond en wanneer het werd gemeld:

- **date_rptd**: datum waarop het misdrijf werd gemeld.
- **date_occ**: datum waarop het misdrijf plaatsvond.
- **time_occ**: tijdstip waarop het misdrijf plaatsvond.

#### 3. Locatiegegevens

Deze variabelen beschrijven waar het misdrijf plaatsvond:

- **area**: nummer van het LAPD-politiegebied.
- **area_name**: naam van het politiegebied.
- **rpt_dist_no**: rapportagedistrict binnen het politiegebied.
- **location**: afgerond adres van het incident.
- **cross_street**: dichtstbijzijnde kruising.
- **lat**: breedtegraad van de locatie.
- **lon**: lengtegraad van de locatie.

#### 4. Misdrijfkenmerken

Deze variabelen beschrijven het gepleegde misdrijf:

- **crm_cd**: code van het misdrijf.
- **crm_cd_desc**: beschrijving van het misdrijf.
- **crm_cd_1**: primaire misdrijfcode.
- **crm_cd_2**, **crm_cd_3**, **crm_cd_4**: eventuele aanvullende misdrijfcodes.
- **mocodes**: modus-operandi-codes die aangeven hoe het misdrijf werd gepleegd.

#### 5. Slachtofferkenmerken

Deze variabelen beschrijven kenmerken van het slachtoffer:

- **vict_age**: leeftijd van het slachtoffer.
- **vict_sex**: geslacht van het slachtoffer.
- **vict_descent**: etnische of culturele achtergrond van het slachtoffer.

#### 6. Overige incidentkenmerken

Deze variabelen geven aanvullende informatie over het incident:

- **premis_cd** en **premis_desc**: type locatie waar het misdrijf plaatsvond.
- **weapon_used_cd** en **weapon_desc**: informatie over een eventueel gebruikt wapen.
- **status** en **status_desc**: status van de zaak.

### Geselecteerde variabelen

Niet alle variabelen zijn relevant voor het onderzoek. Daarom is een selectie gemaakt van de variabelen die het beste aansluiten bij het analyseren van verschillen in criminaliteit tussen gebieden en slachtofferprofielen.

| Variabele | Doel |
|------------|------|
| `date_occ` | Bepalen wanneer een misdrijf plaatsvond |
| `crm_cd_desc` | Het type misdrijf analyseren |
| `part_1_2` | Onderscheid maken tussen ernstigere en minder ernstige misdrijven |
| `area_name` | Criminaliteit tussen verschillende politiegebieden vergelijken |
| `vict_age` | Leeftijdsverschillen tussen slachtoffers onderzoeken |
| `vict_sex` | Verschillen tussen mannelijke en vrouwelijke slachtoffers analyseren |
| `vict_descent` | Verschillen in slachtofferachtergrond onderzoeken |
| `lat`, `lon` | Geografische spreiding van misdrijven visualiseren en hotspots identificeren |

Deze variabelen bevatten samen informatie over wanneer een misdrijf plaatsvond, waar het plaatsvond, welk type misdrijf het betrof en wie het slachtoffer was. Hierdoor vormen ze een goede basis voor het beantwoorden van de onderzoeksvraag.

### Afgeleide variabelen

Naast de oorspronkelijke variabelen zijn twee nieuwe variabelen aangemaakt op basis van `date_occ`:

- **year**: het jaar waarin het misdrijf plaatsvond.
- **month**: de maand waarin het misdrijf plaatsvond.

Deze afgeleide variabelen maken het mogelijk om trends en patronen in criminaliteit over verschillende jaren en maanden te onderzoeken.

## New York

Voor dit onderzoek is gebruikgemaakt van de dataset "NYPD Arrests Data (Historic)", beschikbaar gesteld door de New York City Police Department (NYPD). De dataset bevat informatie over arrestaties die zijn verricht door de politie in New York City. Elke rij in de dataset vertegenwoordigt één arrestatie. De dataset bevat ongeveer 6,26 miljoen observaties (rijen) en 19 variabelen (kolommen).

### Variabelen in de dataset

#### 1. Tijdsgegevens

Deze variabelen beschrijven wanneer de arrestatie plaatsvond:

- **arrest_date**: exacte datum van de arrestatie.

#### 2. Delict- en juridische kenmerken

Deze variabelen beschrijven het delict waarvoor iemand is gearresteerd:

- **pd_cd**: gedetailleerde interne classificatiecode van het delict.
- **pd_desc**: beschrijving van de PD-code.
- **ky_cd**: algemene classificatiecode van het delict.
- **ofns_desc**: beschrijving van het type overtreding of misdrijf.
- **law_code**: verwijzing naar de relevante wetgeving.
- **law_cat_cd**: ernstcategorie van het delict (*felony*, *misdemeanor* of *violation*).

#### 3. Locatiegegevens

Deze variabelen beschrijven waar de arrestatie plaatsvond:

- **arrest_boro**: stadsdeel waarin de arrestatie plaatsvond (*Bronx, Brooklyn, Manhattan, Queens* of *Staten Island*).
- **arrest_precinct**: politiedistrict waarin de arrestatie plaatsvond.
- **jurisdiction_code**: instantie die verantwoordelijk was voor de arrestatie.
- **x_coord_cd**: X-coördinaat binnen het New York State Plane Coordinate System.
- **y_coord_cd**: Y-coördinaat binnen het New York State Plane Coordinate System.
- **latitude**: breedtegraad.
- **longitude**: lengtegraad.
- **lon_lat**: gecombineerd geografisch punt op basis van latitude en longitude.

#### 4. Kenmerken van de verdachte

Deze variabelen beschrijven de persoon die is gearresteerd:

- **age_group**: leeftijdscategorie van de verdachte.
- **perp_sex**: geslacht van de verdachte.
- **perp_race**: etnische of raciale achtergrond van de verdachte.

### Geselecteerde variabelen

Voor deze analyse is een selectie gemaakt van de variabelen die relevant zijn voor het onderzoeken van verschillen tussen de stadsdelen van New York en de kenmerken van arrestaties.

| Variabele | Doel |
|------------|------|
| `arrest_date` | Bepalen wanneer een arrestatie plaatsvond |
| `ofns_desc` | Het type overtreding of misdrijf analyseren |
| `law_cat_cd` | Onderscheid maken tussen overtredingen, misdrijven en ernstige misdrijven |
| `arrest_boro` | Arrestaties tussen de verschillende stadsdelen vergelijken |
| `age_group` | Leeftijdsverschillen tussen verdachten onderzoeken |
| `perp_sex` | Verschillen tussen mannelijke en vrouwelijke verdachten analyseren |
| `perp_race` | Verschillen in achtergrond van verdachten onderzoeken |
| `latitude`, `longitude` | Geografische spreiding van arrestaties in kaart brengen |

Deze variabelen geven gezamenlijk inzicht in wanneer een arrestatie plaatsvond, waar deze plaatsvond, waarvoor iemand werd gearresteerd en welke kenmerken de verdachte had.

### Afgeleide variabelen

Naast de oorspronkelijke variabelen zijn twee nieuwe variabelen aangemaakt op basis van `arrest_date`:

- **year**: het jaar waarin de arrestatie plaatsvond.
- **month**: de maand waarin de arrestatie plaatsvond.

Deze afgeleide variabelen maken het mogelijk om arrestaties over de tijd te analyseren en trends tussen verschillende jaren en maanden te onderzoeken. Hierdoor kan bijvoorbeeld worden bekeken of bepaalde typen misdrijven vaker voorkomen in specifieke perioden of hoe arrestatiepatronen zich in de loop der jaren hebben ontwikkeld.

## vorm van datum netjes gemaakt + kolommen jaar en maand toegevoegd

```python
df_nypd["arrest_date"] = pd.to_datetime(df_nypd["ARREST_DATE"])
df_lapd["DATE OCC"] = pd.to_datetime(df_lapd["DATE OCC"])
```

```python
print(df_nypd["ARREST_DATE"].dropna().iloc[:10].tolist())
print(df_lapd["DATE OCC"].dropna().iloc[:10].tolist())
```

```python
df_nypd["ARREST_DATE"] = pd.to_datetime(
    df_nypd["ARREST_DATE"],
    format="%m/%d/%Y"
)
```

```python
df_nypd["year"] = df_nypd["ARREST_DATE"].dt.year
df_nypd["month"] = df_nypd["ARREST_DATE"].dt.month

df_lapd["year"] = df_lapd["DATE OCC"].dt.year
df_lapd["month"] = df_lapd["DATE OCC"].dt.month
```

```python
df_nypd = df_nypd[
    (df_nypd['ARREST_DATE'].dt.year >= 2020) &
    (df_nypd['ARREST_DATE'].dt.year <= 2023)
]
print(f"NYPD gefilterd: {len(df_nypd)} rijen")
```

```python
df_lapd = df_lapd[
    (df_lapd['DATE OCC'].dt.year >= 2020) &
    (df_lapd['DATE OCC'].dt.year <= 2023)
]

print(f"LAPD gefilterd: {len(df_lapd)} rijen")
```
