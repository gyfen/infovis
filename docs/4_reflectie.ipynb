# Reflectie

# Reflectie op ontvangen feedback
Tijdens de bespreking van onze conceptversie ontvingen wij feedback van zowel medestudenten als de docent assistent. Een van de punten was dat de inleiding nog duidelijker geschreven kon worden. Hoewel het onderwerp en de onderzoeksvragen aanwezig waren, kon de uitleg rondom de visualisaties beter.

Daarnaast kregen wij het advies om de nummering van de figuren consistent toe te passen. In de definitieve versie zijn alle visualisaties voorzien van duidelijke figuurnummers. Hierdoor kunnen lezers eenvoudiger naar specifieke grafieken verwijzen en wordt de structuur van de data story overzichtelijker.

Er werd ook aangegeven dat de bijlagen en verwijzingen hiernaar niet altijd duidelijk waren. Om dit te verbeteren hebben wij de bijlagen opnieuw gestructureerd en de verwijzingen in de tekst aangepast, zodat de gebruikte datasets, aanvullende analyses en bronvermeldingen gemakkelijker terug te vinden zijn.

Terugkijkend heeft de ontvangen feedback ons geholpen om zowel de inhoudelijke kwaliteit als de leesbaarheid van de data story te verbeteren. De structuur is duidelijker geworden, de visualisaties ondersteunen de argumenten beter en de vergelijking tussen New York en Los Angeles is sterker onderbouwd.

# Zelf-reflectie
Over het algemeen verliep de samenwerking goed. De taken waren duidelijk verdeeld en iedereen heeft bijgedragen aan het eindresultaat. Wel hadden we in de beginfase sneller een keuze kunnen maken voor het onderwerp en de datasets, waardoor we meer tijd hadden gehad voor de verdere uitwerking.

Daarnaast had de communicatie rondom het vertrek van onze groepsgenoot Sofia beter gekund. Achteraf gezien hadden we duidelijkere afspraken kunnen maken en mogelijk wat strenger kunnen zijn toen bleek dat zij niet verder zou deelnemen aan het project.

Ondanks deze punten kijken we positief terug op de samenwerking. We hebben elkaar goed aangevuld, gezamenlijk beslissingen genomen en samen gewerkt.
